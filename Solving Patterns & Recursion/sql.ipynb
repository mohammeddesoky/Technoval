```python
import sqlite3

conn = sqlite3.connect("Company.db")
cursor = conn.cursor()
conn.commit()
```

**1- Show all Departments**

```python
query  = """
SELECT * 
FROM Departments
"""
for row in cursor.execute(query):
    print(row)
```

**2- Show all Employees**

```python
query  = """
SELECT * 
FROM Employees
"""
for row in cursor.execute(query):
    print(row)
```

**3- Show all Projects**

```python
query  = """
SELECT * 
FROM Projects
"""
for row in cursor.execute(query):
    print(row)
```

**4- Join Employees with Departments**

```python
cursor.execute("""
SELECT Employees.name, Departments.department_name
FROM Employees
JOIN Departments
ON Employees.department_id = Departments.department_id;
""")
result = cursor.fetchall()
print(result)
```

**5- Count employees per department**

```python
cursor.execute("""
SELECT Departments.department_name, count(*)
FROM Employees
JOIN Departments 
ON Employees.department_id = Departments.department_id
GROUP BY Employees.department_id
""")
result = cursor.fetchall()
print(result)
```

**6- Show Employees Older Than 30**

```python
cursor.execute("""
SELECT name, age
FROM Employees 
WHERE age > 30
""")
result = cursor.fetchall()
print(result)
```

**7- Show Employees With Salary Greater Than 4500**

```python
cursor.execute("""
SELECT name, salary
FROM Employees 
WHERE salary > 4500
""")
result = cursor.fetchall()
print(result)
```

**8- Show Employees and the Projects They Work On**

```python
cursor.execute("""
SELECT Employees.name, Projects.project_name
FROM Employees 
JOIN Employee_Project
ON Employees.employee_id = Employee_Project.employee_id
JOIN Projects
ON Employee_Project.project_id = Projects.project_id
""")
result = cursor.fetchall()
print(result)
```

**9- Show Average Salary Per Department**

```python
cursor.execute("""
SELECT Employees.department_id, Departments.department_name, AVG(Employees.salary) as average_salary
FROM Employees
JOIN Departments 
ON Employees.department_id = Departments.department_id
GROUP BY Employees.department_id
""")
result = cursor.fetchall()
print(result)
```

**10- Show Employees Working on More Than One Project**

```python
cursor.execute("""
SELECT Employees.name , COUNT(Projects.project_id) as project_count
FROM Employees
JOIN Employee_Project 
ON Employees.employee_id = Employee_Project.employee_id
JOIN Projects
ON Employee_Project.project_id = Projects.project_id
GROUP BY Employees.employee_id
HAVING COUNT(Projects.project_id) > 1
""")
result = cursor.fetchall()
print(result)
```
